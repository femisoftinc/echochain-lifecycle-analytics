# Initial Observations

### Laptop Dataset
- Contains laptop specifications and SKU information.
- Will be used as the internal product dataset.

### Warranty Dataset
- Contains warranty claims and component failures.
- Useful for identifying frequently failing components.

### eBay Dataset
- Contains secondary market listings, resale prices, and product conditions.
- Will be matched with internal SKUs to calculate Circularity Score.
